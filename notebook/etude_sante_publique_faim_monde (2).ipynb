Correction du template ci-dessus : la cellule initiale convertissait Valeur avec pd.to_numeric sans argument errors. Elle échouait sur les valeurs "<0.1". Je l'ai supprimée : la cellule suivante fait la conversion avec errors='coerce'.

```python
# Contrôle avant nettoyage : valeurs vides et valeurs "<0.1" pour l'intervalle 2016-2018 (année 2017)
sn_2016_2018 = sous_nutrition.loc[sous_nutrition['Année'] == '2016-2018']
print("Pays pour 2016-2018 :", sn_2016_2018.shape[0])
print("Valeurs vides :", sn_2016_2018['Valeur'].isna().sum())
print("Valeurs '<0.1' :", sn_2016_2018['Valeur'].str.startswith('<', na=False).sum())
```

**Limite du nettoyage :** pour 2016-2018, 99 pays n'ont aucune valeur et 20 ont la valeur « <0.1 » (moins de 0,1 million de personnes). Tous ces cas sont remplacés par 0 (`errors='coerce'` puis `fillna(0)`). Ce choix sous-estime la sous-nutrition : un pays à 0 n'a pas forcément zéro personne sous-alimentée, la donnée est simplement absente ou trop faible pour être précisée.

```python
#Conversion de la colonne (avec l'argument errors=coerce qui permet de convertir automatiquement les lignes qui ne sont pas des nombres en NaN)
sous_nutrition['Valeur'] = pd.to_numeric(sous_nutrition['Valeur'], errors='coerce')
#Puis remplacement des NaN en 0
sous_nutrition['Valeur'] = sous_nutrition['Valeur'].fillna(0)
```

```python
#Combien mange en moyenne un être humain ? Source => PAM/HCR, Guidelines for estimating food and nutritional needs in emergencies
#https://www.unhcr.org/uk/sites/uk/files/legacy-pdf/3b9cbef7a.pdf
#Valeur retenue : 2100 kcal par personne et par jour (valeur de planification, hypothèse simplificatrice)
```

```python
#On commence par faire une jointure entre le data frame population et Dispo_alimentaire afin d'ajouter dans ce dernier la population
pop_2017 = population.loc[population['Année'] == 2017 ,['Zone', 'Population']]

dispo_pop = dispo_alimentaire

dispo_pop = pd.merge(dispo_pop, pop_2017, on='Zone', how='left')
```

**Année retenue :** le lexique précise que le fichier de disponibilité alimentaire concerne l'année 2017. On lui associe donc la population de 2017.

```python
#Affichage du nouveau dataframe
dispo_pop.head()
```

```python
#Création de la colonne dispo_kcal avec calcul des kcal disponibles mondialement
dispo_pop['dispo_kcal']= dispo_pop['Disponibilité alimentaire (Kcal/personne/jour)'] * dispo_pop['Population']
```

```python
#Calcul du nombre d'humains pouvant être nourris
disponibilite = dispo_pop['dispo_kcal'] / 2100



#Calcul et affichage du nombre d'humain pouvant être nourri
print("Le nombre d'humains pouvant être nourri est de {:,} personnes".format(int(disponibilite.sum())))
```

```python
# Graphique : population mondiale et nombre de personnes nourrissables (en milliards)
population_mondiale = pop_2017['Population'].sum() / 1e9
personnes_nourrissables = dispo_pop['dispo_kcal'].sum() / 2100 / 1e9

plt.figure(figsize=(6, 4))
barres = plt.bar(['Population 2017', 'Personnes nourrissables'],
                 [population_mondiale, personnes_nourrissables], color=[GRIS, TEAL])
plt.bar_label(barres, fmt='%.2f Md', padding=5)
plt.ylim(0, 11.5)
plt.ylabel('Milliards de personnes')
plt.title('Population mondiale et personnes nourrissables')
plt.show()
```

**Lecture :** les calories disponibles dans le monde permettraient de nourrir 9,96 milliards de personnes, contre 7,55 milliards d'habitants en 2017 (+ 32 %). C'est un maximum théorique : il ne dit rien de la répartition de la nourriture.

```python
#Calcul de la disponibilité totale
dispo_totale = dispo_pop['Disponibilité intérieure'].sum()
```

```python
#création d'une boucle for pour afficher les différentes valeurs en fonction des colonnes aliments pour animaux, pertes, nourritures, 
colonnes_utilisation = ['Aliments pour animaux', 'Pertes', 'Nourriture','Autres Utilisations', 'Semences', 'Traitement']

for colonne in colonnes_utilisation:
    quantite = dispo_pop[colonne].sum()   
    part = quantite / dispo_totale * 100    
    
    print(colonne, ":", part, "%")
```

```python
# Graphique : part de chaque usage dans la disponibilité intérieure (en %)
colonnes_usage = ['Nourriture', 'Traitement', 'Aliments pour animaux', 'Autres Utilisations', 'Pertes', 'Semences']

parts_usage = {}
for colonne in colonnes_usage:
    parts_usage[colonne] = dispo_pop[colonne].sum() / dispo_totale * 100
parts_usage = pd.Series(parts_usage).sort_values(ascending=False)

ax = parts_usage.plot(kind='barh', color=TEAL, figsize=(8, 4))
ax.invert_yaxis()                      
ax.bar_label(ax.containers[0], fmt='%.1f %%', padding=6)   
ax.margins(x=0.12)
plt.xlabel('Part de la disponibilité intérieure (%)')
plt.title('Répartition de la disponibilité intérieure mondiale')
plt.show()
```

**Lecture :** moins de la moitié (49,5 %) de la disponibilité intérieure mondiale va à l'alimentation humaine. 22,4 % sont transformés, 13,2 % nourrissent les animaux et 4,6 % sont perdus (parts en kg).

### Analyse : utilisation de la disponibilité intérieure mondiale

**Ce que j'ai calculé :** pour chaque utilisation (aliments pour animaux, pertes, nourriture, autres utilisations, semences, traitement), la somme de la colonne dans `dispo_pop` divisée par la disponibilité intérieure totale, multipliée par 100. Les quantités sont en kg.

| Utilisation | Part de la disponibilité intérieure |
|---|---|
| Nourriture | 49,5 % |
| Traitement | 22,4 % |
| Aliments pour animaux | 13,2 % |
| Autres utilisations | 8,8 % |
| Pertes | 4,6 % |
| Semences | 1,6 % |

**Contrôle de cohérence :** les six parts additionnent à 100,1 %. L'écart de 0,1 % est faible, probablement dû aux arrondis ou à de petites incohérences dans les données.

**Conclusion :** moins de la moitié de la disponibilité intérieure mondiale est destinée directement à l'alimentation humaine. Le reste est réparti entre le traitement, l'alimentation animale, d'autres usages, les pertes et les semences.

**Limite :** ces parts sont exprimées en kg, pas en calories, donc elles ne sont pas directement comparables aux résultats de la 3.2.

## Hypothèses et limites

1. **Variables en milliers de tonnes (`dispo_alimentaire`) :** le lexique en annonce 10, j'en compte 11. Hypothèse : 4 colonnes d'acquisition (production, importations, variation de stock, exportations) et 6 d'utilisation font 10, et `Disponibilité intérieure` est le total de l'équation. Contrôle sur les agrumes en Afghanistan : production + importations − exportations = disponibilité intérieure = nourriture + pertes.
2. **Valeurs manquantes de `dispo_alimentaire`** remplacées par 0 : on suppose l'absence de production, d'exportation, etc.
3. **Sous-nutrition :** 99 valeurs vides et 20 valeurs « <0.1 » (2016-2018) sont remplacées par 0, ce qui sous-estime le total et peut exclure des pays très touchés du classement.
4. **Intervalle 2016-2018** retenu pour 2017, car le lexique résume un intervalle de 3 ans par son année du milieu.
5. **Besoin énergétique :** 2 100 kcal par personne et par jour (valeur de planification PAM/HCR, pensée pour l'aide d'urgence). Elle ignore l'âge et l'activité de chaque pays.
6. **Consigne de la 3.4 :** ne précise pas les valeurs à afficher. J'ai calculé la part de chaque utilisation dans la disponibilité intérieure.
7. **Céréales :** liste fondée sur la définition FAO. L'avoine et le sarrasin n'existent pas dans le fichier, peut-être inclus dans « Céréales, Autres ».
8. **Aide alimentaire :** les lignes « Céréales » et « Non-céréales » semblent être des sous-totaux, donc les totaux sont probablement doublés.
9. **Manioc :** la part exportée est calculée par rapport à la production, car la consigne ne précise pas le dénominateur.
10. **RGPD :** les quatre tableaux sont agrégés par pays et par année, aucune ligne ne désigne une personne : il n'y a pas de données personnelles dans les données nettoyées.
